```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df_cien_tec = pd.read_csv("base_indicador-cfp02_2024.csv", header=0, names=["gasto_publico_federal", "gasto_nacional", "porcentaje_fed_nac"], index_col=0)
```

```python
df_cien_tec.head()
```

```python
df_cien_tec.info()
```

```python
df_cien_tec
```

#### El dataframe nos muestra tres indicadores, todos en el periodo 2019-2024.
#### Que nos hablan sobre la inversión en Ciencia, Tecnología e Innovación
#### El primero es el gasto público federal, el segundo el gasto nacional y el tercero el porcentaje que representa el gasto público federal con respecto al nacional. (cifras en MDP)

```python
df_cien_tec.isna().sum()
```

```python
df_cien_tec.duplicated().sum()
```

```python
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(12, 4))
ax[0].boxplot(df_cien_tec['gasto_publico_federal'])
ax[0].set_title("Gasto Public Federal")

ax[1].boxplot(df_cien_tec['gasto_nacional'])
ax[1].set_title("Gasto Nacional")

ax[2].boxplot(df_cien_tec['porcentaje_fed_nac'])
ax[2].set_title("Porcentaje Federal")

plt.tight_layout()
plt.show()
```

#### Los boxplots sugieren que no hay datos atípicos.
