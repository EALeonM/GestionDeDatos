## Operaciones Geométricas

### Para poder realizar operaciones geométricas sobre objetos GeoDataFrame, es IMPORTANTE estar seguros de que el CRS de la geometría es un CRS proyectado para poder hacer cálculos en metros, en sistemas de coordenadas cartesianas, en vez de en ángulos.

#### Aplica .crs sobre el geo df mapa_mx_gj y muestra el resultado

```python
mapa_mx_gj.crs
```

### Si en el primer renglón aparece algo parecido a EPSG:4326 y en los ejes habla de latitud, longitud...entonces pero no está proyectado.  La ubicación de elementos se da en ángulos con un sistema de coordenadas de Longitud y Latitud.  Es necesario convertirlo a un CRS proyectado, como Mercator (EPSG 3857).

### Ejecuta la siguiente celda para realizar la conversión

```python
mapa_mx = mapa_mx_gj.to_crs(epsg=3857)
```

```python
# Observamos nuevamente el crs del nuevo geo df
mapa_mx.crs
```

### Ya se tiene un CRS proyectado, lo que permite especificar coordenadas X,Y en metros.

### Para saber el área (en metros) de cada registro en el geo df se puede aplicar .area como se muestra a continuación (el resultado es una GeoSeries)

```python
mapa_mx.area
```

#### Genera una nueva columna en mapa_mx que contenga el area de cada ENTIDAD expresada en kms (area en metros/1,000,000), muestra los 5 primeros renglones de mapa_mx para comprobar que se realizó ok la tarea

```python
mapa_mx["areakm"] = mapa_mx.area/1000000
```

```python
mapa_mx.head()
```

### Las áreas no son exactas pues las proyecciones Mercator privilegian conservar la forma de los polígonos a expensas de alterar las dimensiones de área.

### Ejercicio:
### Calcular la distancia en kilómetros del Estado Azteca, en la Ciudad de México a los centroides de los Estados.
### Las coordenadas del Estado Azteca son x = -11042932.94161607; y = 2195615.3345249896.
### Para poder hacer la operación, es necesario indicar que esas coordenadas son un objeto geométrico con el mismo CRS que nuestro GeoJSON. Para eespecificar lo anterior se emplea la función Point() de la librería shapely

```python
from shapely.geometry import Point

estadio_azteca = Point(-11042932.94161607,2195615.334524989)
estadio_azteca=gpd.GeoSeries(estadio_azteca, crs=mapa_mx.crs)

# Crear un nueva columna con la distancia 
mapa_mx["distEstAzteca"]=[float(estadio_azteca.distance(centroid))/1000 for centroid in mapa_mx["centroide"]]
```
